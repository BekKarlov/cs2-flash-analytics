```python
blind_df = parser.parse_event("player_blind", player=["team_num"])
blind_df.head(10)
```

```python
import numpy as np

conditions = [
    blind_df["attacker_steamid"] == blind_df["user_steamid"],
    blind_df["attacker_team_num"] == blind_df["user_team_num"],
]

outcomes = ["self", "team", ]

blind_df["flash_type"] = np.select(conditions, outcomes, default="enemy")
blind_df["flash_type"].value_counts()
```

```python

import pandas as pd

flash_summary = pd.crosstab(
    index = [blind_df["entityid"], blind_df["tick"]],
    columns = blind_df["flash_type"],
)

flash_summary.head(10)
```

```python
print(flash_summary.shape)
```

```python
flash_summary["effective"] = flash_summary["enemy"] > 0
flash_summary["effective"].head(10)
```

```python
thrower = blind_df.groupby(["entityid", "tick"])["attacker_name"].first()
thrower.head()
```

```python
flash_summary["thrower"] = thrower
flash_summary.head(10)
```

```python
detonate_df.columns
```

```python
wasted = detonated - blinded

detonate_indexed = detonate_df.set_index(["entityid", "tick"])
wasted_rows = detonate_indexed.loc[list(wasted)]

missing = pd.DataFrame(index=wasted_rows.index)
missing["enemy"] = 0
missing["self"] = 0
missing["team"] = 0
missing["effective"] = False
missing["thrower"] = wasted_rows["user_steamid"]

all_flashes = pd.concat([flash_summary, missing])

print(all_flashes.shape)
```

```python
report = all_flashes.groupby("thrower").agg(
    flashes_thrown = ("effective", "size"),
    effective = ("effective", "sum")
)

report["wasted"] = report["flashes_thrown"] - report["effective"]
report["waste_rate"] = (report["wasted"] / report["flashes_thrown"] * 100).round(1)

report = report.sort_values("waste_rate", ascending=False)
report
```

```python
# One clean lookup: every flash's thrower steamid, from the table that has all 59
thrower_lookup = detonate_df.groupby(["entityid", "tick"])["user_steamid"].first()

# Overwrite thrower on the full table, aligned by (entityid, tick) index
all_flashes["thrower"] = thrower_lookup

# Check: how many unique throwers now?
print(all_flashes["thrower"].nunique())
all_flashes["thrower"].isna().sum()
```

```python
report = all_flashes.groupby("thrower").agg(
    flashes_thrown=("effective", "size"),
    effective=("effective", "sum")
)

report["wasted"] = report["flashes_thrown"] - report["effective"]
report["waste_rate"] = (report["wasted"] / report["flashes_thrown"] * 100).round(1)

report = report.sort_values("waste_rate", ascending=False)
report
```

```python
names = blind_df.groupby("attacker_steamid")["attacker_name"].first()

report.index = report.index.map(names)
report
```

```python
deaths_df = parser.parse_event("player_death", player=["team_num"])
deaths_df.head(10)
```

```python
enemy_blinds = blind_df[blind_df["flash_type"] == "enemy"].copy()

flash_events = enemy_blinds.groupby(["entityid", "tick"]).agg(
    flasher_team = ("attacker_team_num", "first"),
    enemy_team = ("user_team_num", "first")
).reset_index()

flash_events.head(10)
```

```python
parser.parse_header()
```

```python
TICKRATE = 64
WINDOW_SECONDS = 3
WINDOW_TICKS = TICKRATE * WINDOW_SECONDS   # 192
```

```python
test_flash = flash_events.iloc[0]
print(test_flash)
```

```python
# The four conditions, applied across ALL kills at once
converted = deaths_df[
    (deaths_df["attacker_team_num"] == test_flash["flasher_team"]) &
    (deaths_df["user_team_num"] == test_flash["enemy_team"]) &
    (deaths_df["tick"] > test_flash["tick"]) &
    (deaths_df["tick"] <= test_flash["tick"] + WINDOW_TICKS)
]

converted
```

```python
def did_convert(flash):
    matching_kills = deaths_df[
        (deaths_df["attacker_team_num"] == flash["flasher_team"]) &
        (deaths_df["user_team_num"] == flash["enemy_team"]) &
        (deaths_df["tick"] > flash["tick"]) &
        (deaths_df["tick"] <= flash["tick"] + WINDOW_TICKS)
    ]
    return len(matching_kills) > 0
```

```python
flash_events["converted"] = flash_events.apply(did_convert, axis=1)
flash_events.head(10)
```

```python
print("Total flashes:", len(flash_events))
print("Effective flashes that converted to a kill:", flash_events["converted"].sum())
```

```python
enemy_flashes = len(flash_events)            # blinded ≥1 enemy
some_blind = len(flash_summary)              # blinded ≥1 anyone
detonated = len(detonate_df)                 # all flashes

print("Detonated:", detonated)
print("Blinded someone:", some_blind)
print("Blinded an enemy:", enemy_flashes)
print()
print("Blinded nobody:", detonated - some_blind)
print("Blinded only friendly/self (no enemy):", some_blind - enemy_flashes)
```

```python
# thrower_lookup is indexed by (entityid, tick); turn it into a small table to merge
thrower_table = thrower_lookup.reset_index()
thrower_table.columns = ["entityid", "tick", "thrower"]

flash_events = flash_events.merge(thrower_table, on=["entityid", "tick"], how="left")
flash_events.head()
```

```python
conversion_report = flash_events.groupby("thrower").agg(
    enemy_flashes=("converted", "size"),
    converted=("converted", "sum")
)
conversion_report["conversion_rate"] = (
    conversion_report["converted"] / conversion_report["enemy_flashes"] * 100
).round(1)

# swap steamids for names
conversion_report.index = conversion_report.index.map(names)
conversion_report.sort_values("conversion_rate", ascending=False)
```

```python
# Total flashes thrown per player
thrown = all_flashes.groupby("thrower").size()

enemy_hit = flash_events.groupby("thrower")["converted"].agg(["size", "sum"])
enemy_hit.columns = ["enemy_flashes", "converted"]

#Combine
funnel = enemy_hit.join(thrown.rename("total_thrown"), how="right")
funnel = funnel.fillna(0)

#Reordering everything
funnel["total_thrown"] = funnel["total_thrown"].astype(int)
funnel["enemy_flashes"] = funnel["enemy_flashes"].astype(int)
funnel["converted"] = funnel["converted"].astype(int)
funnel["enemy_hit_rate"] = (funnel["enemy_flashes"] / funnel["total_thrown"] * 100).round(1)
funnel["convert_rate_of_thrown"]= (funnel["converted"] / funnel["total_thrown"] * 100).round(1)

funnel.index = funnel.index.map(names)
funnel.sort_values("convert_rate_of_thrown", ascending=False)
```
